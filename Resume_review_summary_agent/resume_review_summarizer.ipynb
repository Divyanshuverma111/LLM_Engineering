# creating small web scraper tool, resume review agent using ollama model

```python
import os
from dotenv import load_dotenv
from web_screape_tool import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
load_dotenv()
def connect_to_ollama(): 
    OLLAMA_BASE_URL = os.getenv("OLLAMA_BASE_URL")
    return OpenAI(base_url=OLLAMA_BASE_URL, api_key="anything")

def chat_with_ollama(messages):
    ollama = connect_to_ollama()
    response = ollama.chat.completions.create(model="llama3.2", messages=messages)
    return response.choices[0].message.content
```

```python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."
system_prompt = """
You are an expert AI Engineering Resume Reviewer with over 10 years of experience hiring, mentoring, and interviewing candidates for AI Engineer, Machine Learning Engineer, Generative AI Engineer, LLM Engineer, Applied AI Engineer, and MLOps Engineer roles.

Your responsibilities are:

Analyze the resume thoroughly.
Evaluate the candidate's suitability for AI Engineering roles.
Identify strengths, weaknesses, and gaps.
Assess technical skills, projects, work experience, achievements, and education.
Suggest specific improvements that increase the candidate's chances of getting shortlisted by recruiters and passing ATS screening.
Identify missing AI/ML, GenAI, LLM, Agentic AI, MLOps, Cloud, and Software Engineering skills that are expected in today's market.
Highlight vague, weak, or generic bullet points and rewrite them using strong action-oriented language.
Recommend measurable achievements and impact-driven statements wherever possible.
Evaluate whether projects demonstrate real-world AI engineering skills.
Provide actionable recommendations prioritized by impact.

When reviewing:

Be constructive, practical, and honest.
Focus on industry expectations for AI Engineer roles in top product companies and AI startups.
Consider ATS optimization.
Consider technical depth, business impact, and relevance to modern AI engineering roles.
Do not invent candidate achievements.
Base recommendations only on the provided resume content.

Respond in Markdown.

Output format:

Resume Review Summary
Overall Assessment
Strengths
Weaknesses
ATS Score (0-100)
AI Engineer Role Readiness
Current Level
Suitable Roles
Missing Skills
Section-by-Section Review
Professional Summary
Experience
Projects
Skills
Education
High Priority Improvements
Suggested Resume Bullet Rewrites
Final Recommendations

"""

# Define our user prompt
user_prompt_prefix = """
You are reviewing the following resume for AI Engineering positions.

Analyze the resume and provide:

Overall resume assessment.
ATS optimization review.
AI Engineer role readiness evaluation.
Strengths and weaknesses.
Missing technical skills and technologies.
Project quality assessment.
Experience quality assessment.
Suggestions for improving recruiter appeal.
Suggestions for improving technical credibility.
Specific bullet-point rewrites where applicable.
A prioritized action plan to improve the resume.

Resume Content:

"""

# preview of calling frontier model using ollama
def user_input_to_messages(website):
# See how this function creates exactly the format above
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

messages = user_input_to_messages("https://divyanshu-verma1.github.io/Portfolio/")
response = chat_with_ollama(messages)
print(response)
```

# web scraper agent

```python
#trying the utility function to fetch website content
url = "https://en.wikipedia.org/wiki/RAG"
website_content = fetch_website_contents(url)
print(website_content)
```

```python
# And now: call the OpenAI API. You will get very familiar with this!
def summarize(url):
    website = fetch_website_contents(url)
    openai = connect_to_ollama()
    response = openai.chat.completions.create(
        model = "llama3.1",
        messages = user_input_to_messages(website)
    )
    return response.choices[0].message.content
```

```python
summarize("https://divyanshuverma111.github.io/Divyanshu_portfolio/")
```

```python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://divyanshu-verma1.github.io/Portfolio/")
```
